```python
# Data  Visualization

import matplotlib.pyplot as plt

import seaborn as sns

import numpy as np
```

```python
# Show the Relationship between Sales person and Totalprice in Bar Diagram
plt.xlabel('Salesperson')
plt.ylabel('TotalPrice')
plt.bar(Product_Sales_Region['Salesperson'],Product_Sales_Region['TotalPrice'], color='Red')
plt.title(' Salesperson TotalPrice Relation')
plt.show()
```

```python
# Findings

Above the diagram Salesperson 'Frank' has highest TotalPrice and 'Carios' has lowest TotalPrice
```

```python
# Show Relation between Salesperson and TotalPrice in Line Chart

data=Product_Sales_Region[['Product','Salesperson','TotalPrice']]

df = pd.DataFrame(data)

target_Product ='Laptop'

filtered_df = df[df['Product'] == target_Product]

plt.plot(filtered_df['Salesperson'], filtered_df['TotalPrice'], marker='o', linestyle='-', color='blue')

plt.xlabel('Salesperson')

plt.ylabel('TotalPrice')

plt.title('TotalPrice Trend for {Laptop}')

plt.legend()

plt.grid(True)

plt.show()
```

```python
# Findings

In this graph Alice is top person and Eva is Lowest person in TotalPrice
```

```python
#Show Ralation Between Salesperson and TotalPrice in Pie diagram

plt.pie(Product_Sales_Region['TotalPrice'],labels=Product_Sales_Region['Salesperson'], autopct='%1.1f%%')
plt.title('Salesperson TotalPrice Realation')
plt.show()
```

```python
#show TotalPrice price distibution in histogram

data=Product_Sales_Region['TotalPrice']
plt.hist(data,bins=30,color='skyblue',edgecolor='Black')
plt.xlabel('TotalPrice')
plt.ylabel('Frequency')
plt.title(' TotalPrice Distribution')
plt.show()
```

```python
# Findings

From this Graph the Freequancy of TotalPrice is decrease when Totalprice incresed
```

```python
#show TotalPrice Distribtion in Boxplot

data=Product_Sales_Region['TotalPrice']
sns.boxplot(x=data)
plt.title('TotalPrice Distribution')
plt.show()
```

```python
#Show Product TotalPrice Distribution using Scatter chart

data=Product_Sales_Region[['TotalPrice','Product']]

plt.scatter(data['TotalPrice'],data['Product'])
plt.title("Product TotalPrice Distribution")
plt.xlabel('TotalPrice')
plt.ylabel('Product')
plt.show()
```

# Findings

In scatter chart the product 'Phone' is high TotalPrice and Desk is lowest

```python
#Show Product TotalPrice Distribution using Heatmap

pivot_data=Product_Sales_Region.pivot_table(values='TotalPrice',index='Product',columns='Region')

plt.figure(figsize=(15,8))

sns.heatmap(pivot_data, annot=True,fmt=".0f", cmap="YlGnBu")

plt.title("Product Product Distribution")

plt.show()
```

```python
# Findings

In Heatmap, Printer in East region is high TotalPrice(3635) and Phone in North region is Lowest(1941)
```

```python
##Show Product TotalPrice Distribution using Lolipop Plot

data=Product_Sales_Region[['Product','TotalPrice']]

x=data['Product']

y=data['TotalPrice']

plt.stem(x, y, linefmt='grey', markerfmt='D', basefmt=" ")

plt.title("Product TotalPrice Distribution")

plt.xlabel("Product")

plt.ylabel("TotalPrice")

plt.show()
```

```python
# Findings
From this figure 'Phone' is high TotalPrice and Desk is Lowest
```

```python
# Show Product TotalPrice Distribution using Area Chart
data=Product_Sales_Region[['Product','TotalPrice']]
x=data['Product']
y=data['TotalPrice']
plt.figure(figsize=(8, 5))
plt.fill_between(x, y, color="skyblue", alpha=0.4)
plt.plot(x, y, color="Slateblue", alpha=0.6, linewidth=2)
plt.title("Product TotalPrice Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Product")
plt.ylabel("TotalPrice")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Findings
From this figure the product Phone is the high TotalPrice and Chair is Lowest

```python
# Show Salesperson TotalPrice Distribution using Violin Plot
data1=Product_Sales_Region[['Region','TotalPrice']]
sns.violinplot(data=data1, x="Region", y="TotalPrice", palette="muted")
plt.title("Region TotalPrice Distribution")
plt.xlabel("Region")
plt.ylabel("TotalPrice")
plt.show()
```

# Findings
Totalprice in East region is drastically decrease morethan 4000 compared to other regions
